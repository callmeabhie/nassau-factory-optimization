# STEP 3: EXPLORATORY DATA ANALYSIS (EDA)

```python
# Load the clean dataset.
try:
    df_eda = df_clean.copy()
except NameError:
    df_eda = df.copy()

sns.set_theme(style="whitegrid")

print("=" * 60)
print("STEP 3: EXPLORATORY DATA ANALYSIS")
print("=" * 60)

print("Dataset shape:", df_eda.shape)
print("Unique orders:", df_eda["Order ID"].nunique())
print("Unique customers:", df_eda["Customer ID"].nunique())
print("Unique products:", df_eda["Product Name"].nunique())
```

```python
print("\n2. NUMERICAL SUMMARY")

display(
    df_eda[
        ["Sales", "Units", "Gross Profit", "Cost"]
    ].describe()
)
```

```python
print("\n3. SHIPPING MODE ANALYSIS")

shipping_counts = df_eda["Ship Mode"].value_counts()

display(shipping_counts.rename("Record Count").to_frame())

plt.figure(figsize=(9, 5))
sns.countplot(data=df_eda, x="Ship Mode", order=shipping_counts.index)
plt.title("Records by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Number of Records")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
print("\n4. REGIONAL SALES AND PROFIT")

region_summary = (
    df_eda.groupby("Region", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum"),
        Total_Units=("Units", "sum"),
        Record_Count=("Order ID", "size")
    )
    .sort_values("Total_Sales", ascending=False)
)

display(region_summary)

plt.figure(figsize=(9, 5))
sns.barplot(data=region_summary, x="Region", y="Total_Sales")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 5))
sns.barplot(data=region_summary, x="Region", y="Total_Gross_Profit")
plt.title("Total Gross Profit by Region")
plt.xlabel("Region")
plt.ylabel("Gross Profit")
plt.tight_layout()
plt.show()
```

```python

print("\n5. TOP 10 PRODUCTS BY SALES")

product_sales = (
    df_eda.groupby("Product Name", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Units=("Units", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .head(10)
)

display(product_sales)

plt.figure(figsize=(11, 6))
sns.barplot(data=product_sales, x="Total_Sales", y="Product Name")
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
print("\n6. TOP 10 PRODUCTS BY GROSS PROFIT")

product_profit = (
    df_eda.groupby("Product Name", as_index=False)
    .agg(Total_Gross_Profit=("Gross Profit", "sum"))
    .sort_values("Total_Gross_Profit", ascending=False)
    .head(10)
)

display(product_profit)

plt.figure(figsize=(11, 6))
sns.barplot(
    data=product_profit,
    x="Total_Gross_Profit",
    y="Product Name"
)
plt.title("Top 10 Products by Gross Profit")
plt.xlabel("Gross Profit")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
print("\n7. SALES BY DIVISION")

division_sales = (
    df_eda.groupby("Division", as_index=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Gross_Profit=("Gross Profit", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
)

display(division_sales)

plt.figure(figsize=(8, 5))
sns.barplot(data=division_sales, x="Division", y="Total_Sales")
plt.title("Sales by Division")
plt.xlabel("Division")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
print("\n8. TOP 10 PRODUCTS BY UNITS SOLD")

product_units = (
    df_eda.groupby("Product Name", as_index=False)["Units"]
    .sum()
    .sort_values("Units", ascending=False)
    .head(10)
)

display(product_units)

plt.figure(figsize=(11, 6))
sns.barplot(data=product_units, x="Units", y="Product Name")
plt.title("Top 10 Products by Units Sold")
plt.xlabel("Total Units")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python

print("\n9. SALES VS GROSS PROFIT")

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df_eda,
    x="Sales",
    y="Gross Profit",
    hue="Division",
    alpha=0.7
)
plt.title("Sales vs Gross Profit")
plt.xlabel("Sales")
plt.ylabel("Gross Profit")
plt.tight_layout()
plt.show()
```

```python
print("\n10. CORRELATION ANALYSIS")

numeric_df = df_eda[
    ["Sales", "Units", "Gross Profit", "Cost"]
].select_dtypes(include=np.number)

correlation = numeric_df.corr()

display(correlation)

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
print("\n11. OVERALL FINANCIAL SUMMARY")

total_sales = df_eda["Sales"].sum()
total_profit = df_eda["Gross Profit"].sum()
total_cost = df_eda["Cost"].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Gross Profit:", round(total_profit, 2))
print("Total Cost:", round(total_cost, 2))

if total_sales != 0:
    margin = (total_profit / total_sales) * 100
    print("Gross Profit Margin:", round(margin, 2), "%")
```

```python
print("\n12. BUSINESS SUMMARY")
print("Most common shipping mode:",shipping_counts.index[0])
print("Region with highest sales:",region_summary.iloc[0]["Region"])
print("Product with highest sales:",product_sales.iloc[0]["Product Name"])
print("Product with highest gross profit:",product_profit.iloc[0]["Product Name"])
print("\nSTEP 3 EDA COMPLETED!")
```

# STEP 4: FEATURE ENGINEERING

```python
try:
    df_fe = df_clean.copy()
except NameError:
    df_fe = df.copy()

print("=" * 60)
print("STEP 4: FEATURE ENGINEERING")
print("=" * 60)

print("Original Dataset Shape:", df_fe.shape)
```

```python
df_fe["Order Date"] = pd.to_datetime(df_fe["Order Date"], errors="coerce")
df_fe["Ship Date"] = pd.to_datetime(df_fe["Ship Date"],errors="coerce")
print("\nDATE VALIDATION")
print("Invalid Order Dates:",df_fe["Order Date"].isna().sum())
print("Invalid Ship Dates:",df_fe["Ship Date"].isna().sum())
```

```python
df_fe["Order_Year"] = df_fe["Order Date"].dt.year
df_fe["Order_Month"] = df_fe["Order Date"].dt.month
df_fe["Order_Quarter"] = ( df_fe["Order Date"].dt.quarter)
df_fe["Order_Day"] = (df_fe["Order Date"].dt.day)
df_fe["Order_DayOfWeek"] = ( df_fe["Order Date"].dt.dayofweek)
print("\nORDER DATE FEATURES CREATED")
```

```python
valid_dates = (df_fe["Order Date"].notna()& df_fe["Ship Date"].notna())
df_fe["Lead_Time_Days"] = np.nan
df_fe.loc[valid_dates, "Lead_Time_Days"] = (df_fe.loc[valid_dates, "Ship Date"]- df_fe.loc[valid_dates, "Order Date"]).dt.days
print("\nLEAD TIME VALIDATION")
print("Valid Date Pairs:", valid_dates.sum())
print("Ship Date Before Order Date:",(df_fe.loc[valid_dates, "Lead_Time_Days"] < 0).sum())
print("Lead Time Summary:")
display(df_fe["Lead_Time_Days"].describe())
```

```python
df_fe["Profit_Margin_Pct"] = np.where(df_fe["Sales"] != 0,(df_fe["Gross Profit"] / df_fe["Sales"]) * 100,np.nan)
print("\nPROFIT MARGIN FEATURE CREATED")
```

```python
df_fe["Sales_Per_Unit"] = np.where(df_fe["Units"] != 0,df_fe["Sales"] / df_fe["Units"],np.nan)
print("Sales Per Unit Feature Created")
```

```python
df_fe["Cost_Per_Unit"] = np.where(df_fe["Units"] != 0,df_fe["Cost"] / df_fe["Units"],np.nan)
print("Cost Per Unit Feature Created")
```

```python
df_fe["Profit_Per_Unit"] = np.where(df_fe["Units"] != 0,df_fe["Gross Profit"] / df_fe["Units"],np.nan)
print("Profit Per Unit Feature Created")
```

```python
df_fe["Order_Size_Category"] = pd.cut(df_fe["Units"],bins=[-np.inf, 1, 5, 10, np.inf],labels=["Small","Medium","Large","Very Large"])
print("\nORDER SIZE CATEGORY CREATED")
```

```python
df_fe["Shipping_Mode_Category"] = (df_fe["Ship Mode"].astype("category"))
print("Shipping Mode Category Created")
```

```python
print("\n" + "=" * 60)
print("NEW FEATURE PREVIEW")
print("=" * 60)
new_features = [
    "Order_Year",
    "Order_Month",
    "Order_Quarter",
    "Order_DayOfWeek",
    "Lead_Time_Days",
    "Profit_Margin_Pct",
    "Sales_Per_Unit",
    "Cost_Per_Unit",
    "Profit_Per_Unit",
    "Order_Size_Category",
    "Shipping_Mode_Category"
]
display(df_fe[new_features].head(10))
```

```python
print("\nFEATURE QUALITY CHECK")
print(df_fe[new_features].isnull().sum())
```

```python
print("\n" + "=" * 60)
print("STEP 4 SUMMARY")
print("=" * 60)
print("Original Rows:", len(df))
print("Feature-Engineered Rows:", len(df_fe))
print("Total Columns:", df_fe.shape[1])
print("\nAll Columns:")
print(df_fe.columns.tolist())
```

```python
df_fe.to_csv("nassau_feature_engineered.csv",index=False)
print("\nDataset saved as:")
print("nassau_feature_engineered.csv")
print("\nSTEP 4 COMPLETED!")
```
