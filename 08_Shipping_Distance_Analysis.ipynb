```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("Nassau Candy Distributor.csv")
factories = pd.read_csv("factory_locations.csv")
print("Dataset loaded:", df.shape)
print("Factories loaded:", factories.shape)
```

```python
possible_coordinate_columns = [
    col for col in df.columns
    if any(word in col.lower()
           for word in ["latitude", "longitude", "lat", "lon"])
]

print("Possible coordinate columns:", possible_coordinate_columns)

if not possible_coordinate_columns:
    print(
        "Customer coordinates are not present. "
        "We need verified coordinates before calculating distances."
    )
```

```python
customer_locations = (
    df[["Country/Region", "State/Province", "City", "Postal Code"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Unique customer locations:", len(customer_locations))
display(customer_locations.head(20))
```

```python
customer_locations.to_csv(
    "customer_locations_for_geocoding.csv",
    index=False
)
print("Customer location list saved successfully.")
```
